```python
import pandas as pd
import time
from geopy.geocoders import Nominatim

df = pd.read_csv("/Users/wusiyi/Documents/CASA/dis/DATA/soho_loading_bays.csv")

geolocator = Nominatim(user_agent="soho_dissertation_research")

def geocode_street(street_name):
    """先用街道名本身做粗定位,取街道中点作为近似坐标"""
    try:
        location = geolocator.geocode(f"{street_name}, Soho, London, UK")
        time.sleep(1)  # Nominatim有请求频率限制,必须sleep
        if location:
            return location.latitude, location.longitude
    except Exception as e:
        print(f"Error geocoding {street_name}: {e}")
    return None, None

df['lat'], df['lon'] = zip(*df['street'].apply(geocode_street))
print(df[['street', 'lat', 'lon']])
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'street': ['Frith Street', 'Dean Street', 'Greek Street', 
               'Kingly Street', 'Kingly Street', 'Kingly Street', 
               'Kingly Street', 'Kingly Street', 
               'Marshall Street', 'Rupert Street'],
    'lat': [51.514769, 51.515142, 51.514265, 
            51.512662, 51.512662, 51.512662, 51.512662, 51.512662,
            51.512902, 51.512214],
    'lon': [-0.132311, -0.133475, -0.131016,
            -0.139314, -0.139314, -0.139314, -0.139314, -0.139314,
            -0.137594, -0.133633]
})

# 给Kingly Street的重复点做一个小的纵向偏移(沿街道走向近似用纬度微调代替)
kingly_mask = df['street'] == 'Kingly Street'
n_kingly = kingly_mask.sum()
offsets = np.linspace(-0.0006, 0.0006, n_kingly)  # 约±65米范围内展开
df.loc[kingly_mask, 'lat'] = df.loc[kingly_mask, 'lat'].values + offsets

print(df)
```

```python
import geopandas as gpd
from shapely.geometry import Point

df['geometry'] = df.apply(lambda row: Point(row['lon'], row['lat']), axis=1)
loading_bays_gdf = gpd.GeoDataFrame(df, geometry='geometry', crs="EPSG:4326")

loading_bays_gdf.to_file(
    "/Users/wusiyi/Documents/CASA/dis/DATA/raw/soho_loading_bays.geojson",
    driver="GeoJSON"
)

print(f"共保存 {len(loading_bays_gdf)} 个装卸区点")
```
