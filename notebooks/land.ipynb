```python
import sys
sys.path.insert(0, "/Users/wusiyi/Documents/CASA/dis/DATA/files")

from environment import load_environment
import json

DATA_DIR = "/Users/wusiyi/Documents/CASA/dis/DATA"
env = load_environment(DATA_DIR)

# 导出每个节点的坐标 + 两个密度值
nodes_data = []
for node, (x, y) in env.node_xy.items():
    nodes_data.append({
        "node": int(node),
        "x": x,
        "y": y,
        "land_use_mix": env.land_use_mix.get(node, 0.0),
        "activity_density": env.activity_density.get(node, 0.0),
    })

# 导出路网边（用于画底图路网线条）
edges_data = []
for u, v in env.graph.edges():
    if u in env.node_xy and v in env.node_xy:
        x0, y0 = env.node_xy[u]
        x1, y1 = env.node_xy[v]
        edges_data.append([x0, y0, x1, y1])

# 导出停车点坐标
parking_coords = []
for node in env.cycle_parking_nodes:
    if node in env.node_xy:
        x, y = env.node_xy[node]
        parking_coords.append([x, y])

with open("/Users/wusiyi/Documents/CASA/dis/DATA/map_export.json", "w") as f:
    json.dump({
        "nodes": nodes_data,
        "edges": edges_data,
        "parking": parking_coords,
    }, f)

print(f"节点数: {len(nodes_data)}, 边数: {len(edges_data)}, 停车点数: {len(parking_coords)}")
```
