```python
import pandas as pd
import math

df = pd.read_csv('/Users/wusiyi/Documents/CASA/dis/DATA/dft_countpoints_local_authority_id_109.csv')

lat0, lon0 = 51.5135, -0.1365  # 你的Soho圆心
R = 6371000.0

def haversine(lat, lon):
    lat0r = math.radians(lat0)
    dlat = math.radians(lat - lat0)
    dlon = math.radians(lon - lon0)
    a = math.sin(dlat/2)**2 + math.cos(lat0r)*math.cos(math.radians(lat))*math.sin(dlon/2)**2
    return 2 * R * math.asin(math.sqrt(a))

df["dist_to_center_m"] = df.apply(lambda r: haversine(r["latitude"], r["longitude"]), axis=1)

nearby = df[df["dist_to_center_m"] <= 1900].sort_values("dist_to_center_m")
print(nearby[["count_point_id", "road_name", "road_type", "dist_to_center_m"]].to_string(index=False))
```
